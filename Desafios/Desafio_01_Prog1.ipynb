#Parte 1: Variáveis, Matemática e Lógica
Nesta etapa, o seu programa **não** vai interagir com o usuário (não use input()). Você mesmo vai criar as variáveis com os valores que a história forneceu, fazer os cálculos e mandar o computador mostrar o resultado.

1. O Orçamento da Confraternização

    O setor de Recursos Humanos está organizando uma festa. Eles já sabem que terão exatamente 45 convidados e que o buffet cobra R$ 38,50 por pessoa.
    
    Tarefa: Crie uma variável chamada convidados e guarde o número 45. Crie uma variável chamada custo_pessoa e guarde o valor 38.50. Crie uma terceira variável que multiplique as duas primeiras para descobrir o total. Use o print() para exibir o custo total da festa.

2. O Desempenho da Frota

    Um carro da frota da empresa fez uma viagem. No painel, constam as seguintes informações fixas: ele percorreu 320 quilômetros e consumiu 25 litros de combustível. A meta da empresa é que os carros façam, no mínimo, 12 km por litro.

    Tarefa: Crie as variáveis para a distância e para os litros. Crie uma variável rendimento que divide a distância pelos litros. Por fim, crie uma expressão relacional para verificar se o rendimento alcançou a meta (maior ou igual a 12) e imprima o resultado (True ou False).

3. O Cofre Duplo de Segurança

    O cofre de um banco exige que dois gerentes girem suas chaves ao mesmo tempo. O painel eletrônico lê a Chave A e a Chave B. O sistema diz que a Chave A foi girada (True), mas a Chave B não foi (False).

    Tarefa: Crie uma variável chave_a valendo True e uma chave_b valendo False. Crie uma terceira variável que use o operador lógico adequado ao problema para checar se as duas chaves estão giradas simultaneamente. Mande o computador imprimir o resultado final da abertura do cofre.

4. A Conversão de Câmbio em Tempo Real

    Um importador precisa comprar equipamentos em Dólares, mas a cotação muda todo dia. Ele precisa de um programa rápido para fazer a conversão para Reais.

    Tarefa: Use o input() para perguntar: "Qual a cotação do dólar hoje?". Depois, pergunte: "Quantos dólares você precisa comprar?". Multiplique os dois valores e use o print() para mostrar a frase: "Você precisará gastar R$ [resultado] Reais".

5. A Divisão da Conta com Gorjeta

    Um grupo de amigos foi a um restaurante. Eles querem dividir o valor total da conta igualmente, mas antes da divisão, eles sempre adicionam 10% de gorjeta para o garçom.

    Tarefa: Pergunte o valor da conta e pergunte quantas pessoas estão na mesa. Em seguida, pegue o valor da conta e multiplique por 1.10 (para adicionar a gorjeta). Depois, divida o novo valor pelo número de pessoas. Imprima o valor final que cada um deve pagar.

6. A Avaliação Automática de Meta

    Um gerente de vendas quer um verificador rápido. Ele digita qual era a meta de vendas da loja e, depois, digita quanto a loja realmente vendeu. O programa deve dizer se a meta foi batida.

    Tarefa: Pergunte ao usuário o valor da meta e pergunte o valor real vendido. Faça o computador comparar se o valor vendido é maior ou igual à meta e imprima o resultado. (A tela deve mostrar apenas True ou False).

7. O Radar Inteligente de Rodovia

    A concessionária da rodovia quer automatizar os painéis eletrônicos. O limite de velocidade é 80 km/h. Se o motorista passar acima do limite, o painel deve dar um esporro. Se passar abaixo, deve agradecer.
    
    Tarefa: Peça para o motorista digitar a velocidade atual do veículo.

    Use um if para checar se a velocidade é estritamente maior que 80. Se for, imprima: "MULTADO! Reduza a velocidade imediatamente!".

    Use outro if para o cenário contrário e imprima: "Boa viagem! Continue respeitando as leis de trânsito."

8. O Analista de Crediário Automático

    Uma loja de eletrodomésticos vende produtos parcelados em 10 vezes sem juros. Porém, a loja tem uma regra financeira de proteção: o valor da parcela mensal não pode comprometer mais do que 30% do salário do cliente.

    Tarefa:

    1. Pergunte o valor do produto que o cliente quer comprar.

    2. Pergunte o valor do salário mensal do cliente.

    3. Calcule o valor da parcela (valor do produto dividido por 10).

    4. Calcule o limite seguro do cliente de 30% do salario (salário multiplicado por 0.30).

    5. Use um if para comparar: se a parcela for menor ou igual ao limite seguro, imprima "Crédito Aprovado! Parabéns pela compra".

    6. Use outro if para bloquear a venda: imprima "Crédito Negado! O valor da parcela compromete sua renda."
