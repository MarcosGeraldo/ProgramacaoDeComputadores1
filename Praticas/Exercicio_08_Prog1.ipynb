#Exercícios: Laços de Repetição (while)

## Questão 1: Acúmulo Progressivo de Carbono em Chaminé de Exaustão

Em um forno de reaquecimento, **resíduos de carbono** e **fuligem** impregnam a parede interna da chaminé de exaustão. À medida que a camada inicial se fixa, a rugosidade aumenta e a retenção de fuligem se **agrava a cada dia de operação**: no **$1º$** dia acumula-se **$1\text{ mm}$** de fuligem, no **$2º$** dia acumulam-se **$2\text{ mm}$**, no **$3º$** dia **$3\text{ mm}$**, e assim sucessivamente.

Escreva um programa em Python que leia via input() o **total de dias de operação** contínua **$N$** (inteiro). Utilizando o laço while, calcule a **espessura total acumulada** da camada de carbono ao final dos **$N$** dias.

**Fórmula do Acúmulo**: $S = \sum_{k=1}^{N} k = 1 + 2 + 3 + \dots + N$

**Entrada**: Um número inteiro **$N$** indicando os dias de operação (ex: 10).

**Saída**: A espessura total acumulada da camada de carbono em mm (ex: para $N = 10$, a saída será $55\text{ mm}$).

##Questão 2: Tempo de Aquecimento Linear em Forno

Uma chapa metálica entra em um forno à temperatura inicial de **$20,0\text{ °C}$**. A temperatura da chapa aumenta exatamente **$25,0\text{ °C}$** a cada minuto de permanência no forno.

Escreva um programa em Python que receba a **temperatura alvo desejada** **$T_{\text{alvo}}$** (float). Utilizando o laço while, calcule quantos minutos são necessários para que a temperatura da chapa atinja ou ultrapasse $T_{\text{alvo}}$.

**Atualização**: $T_{\text{atual}} = T_{\text{atual}} + 25.0$

**Entrada**: Temperatura alvo em °C (ex: 1100.0).

**Saída**: O tempo total em minutos e a temperatura final atingida.

##Questão 3: Tempo de Resfriamento Linear

Uma peça fundida sai do molde à temperatura de **$900,0\text{ °C}$** e perde exatamente **$12,5\text{ °C}$** por minuto durante o resfriamento ao ar.

Escreva um programa em Python que receba a temperatura limite de segurança **$T_{\text{limite}}$** (float). Utilizando o laço while, diminua a temperatura em **$12,5\text{ °C}$** por minuto até que ela fique menor ou igual a **$T_{\text{limite}}$**.

**Atualização**: $T_{\text{atual}} = T_{\text{atual}} - 12.5$

**Entrada**: Temperatura limite em °C (ex: 150.0).

**Saída**: O tempo decorrido em minutos e a temperatura final da peça.



##Questão 4: Perda de Espessura por Corrosão

A parede de uma tubulação de aço possui espessura inicial de **$20,0\text{ mm}$**. Devido à ação corrosiva de um fluido, a parede perde **$5\%$** da sua espessura a cada ano de operação.

Crie um programa em Python que leia a **espessura mínima de segurança** $e_{\text{min}}$ (float). Utilizando o laço while, calcule a redução sucessiva da espessura **ano a ano** até que ela se torne menor que $e_{\text{min}}$.

**Atualização**: $e_{\text{nova}} = e_{\text{atual}} \times 0.95$

**Entrada**: Espessura limite em mm (ex: 8.0).

**Saída**: O número total de anos e a espessura final da parede.

##Questão 5: Crescimento de Grão com Taxa Acelerada

No recozimento de uma liga metálica, o tamanho inicial do grão é **$10,0\,\mu\text{m}$**. Na primeira hora, o grão cresce **$2,0\,\mu\text{m}$**. A cada hora subsequente, o ganho de tamanho aumenta em **$1,5\,\mu\text{m}$** em relação à hora anterior (ou seja: ganha $2,0\,\mu\text{m}$ na 1ª hora, $3,5\,\mu\text{m}$ na 2ª hora, $5,0\,\mu\text{m}$ na 3ª hora, e assim por diante).

Crie um programa que leia o **tamanho de grão alvo** $G_{\text{alvo}}$ (float). Utilizando o laço while, calcule quantas horas são necessárias para o grão atingir ou ultrapassar essa dimensão.

**Atualização do passo**: $v_{\text{novo}} = v_{\text{atual}} + 1.5$

**Atualização do tamanho**: $G_{\text{novo}} = G_{\text{atual}} + v_{\text{novo}}$

**Entrada**: Tamanho limite em $\mu\text{m}$ (ex: 50.0).

**Saída**: Quantidade de horas e tamanho final do grão.

##Questão 6: Balanço de Massa Acumulado em Panela

Para ajustar a composição de uma liga de aço, o operador adiciona cargas de liga na panela. O processo exige atingir uma massa acumulada mínima $M_{\text{meta}}$.

Escreva um programa em Python que leia o valor de **$M_{\text{meta}}$** (float). Em seguida, utilize o laço while para **solicitar sucessivamente a massa de cada adição** (float) via input(), acumulando o total até que a massa acumulada seja maior ou igual a $M_{\text{meta}}$.

**Entrada**: Massa meta $M_{\text{meta}}$ e, em seguida, os valores numéricos de cada adição.

**Saída**: Quantidade de adições realizadas e a massa total final acumulada.


##Questão 7: Classificação de Dureza de Lote (Comparações Internas)

Um lote de $N$ corpos de prova deve passar por teste de dureza. A faixa de especificação técnica é de $180$ a $220$ (inclusive).

Escreva um programa em Python que **leia o total de amostras** $N$ (inteiro). Utilizando o laço while, leia a dureza de cada uma das **$N$** amostras e **conte** quantas estão dentro da faixa especificada (Aprovadas) e quantas estão fora (Rejeitadas).

**Entrada**: Quantidade $N$ e, na sequência, os $N$ valores de dureza (floats).

**Saída**: Total de peças aprovadas e total de peças rejeitadas.

##Questão 8: Desgaste Acelerado de Refratário por Taxa Composta

O revestimento refratário de um forno possui espessura inicial de **$300,0\text{ mm}$**. No primeiro mês de operação, o desgaste é de **$10,0\text{ mm}$**. A cada mês seguinte, a taxa de desgaste aumenta em **$8\%$** em relação ao desgaste do mês anterior devido ao ataque químico do forno.

Escreva um programa em Python que **leia a espessura crítica do revestimento** $e_{\text{critica}}$ (float). Utilizando o laço while, **simule o desgaste mês a mês** enquanto a espessura residual for maior ou igual a $e_{\text{critica}}$.

**Atualização da taxa de desgaste**: $d_{\text{novo}} = d_{\text{atual}} \times 1.08$

**Atualização da espessura**: $e_{\text{nova}} = e_{\text{atual}} - d_{\text{novo}}$

**Entrada**: Espessura crítica em mm (ex: 100.0).

**Saída**: Meses de operação suportados e a espessura residual final.
