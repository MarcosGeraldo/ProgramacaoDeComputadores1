
# Guia de Exercícios Práticos: Expressões e Operadores no Google Colab

**Como funcionará a atividade?**
Em cada exercício, você lerá uma breve situação do mundo real. Sua tarefa é traduzir essa situação para a "linguagem do computador" criando uma única expressão usando os operadores que aprendemos. Você digitará essa expressão em uma célula de código do Google Colab e executará (pressionando Shift + Enter) para que o computador calcule a resposta ou avalie se a situação é Verdadeira (True) ou Falsa (False).



Exemplo Inicial:
João foi à padaria com uma nota de 20 reais. Ele comprou 4 pães de queijo, sendo que cada um custa 3 reais. Quanto ele receberá de troco?

O que você deve digitar na célula do Colab:
20 - (4 * 3)

Para avaliar o resultado basta clicar no botão "play" que fica a esquerda da célula de código.

Resultado que o Colab irá mostrar:
8

```python
20 - (4 * 3)
```
