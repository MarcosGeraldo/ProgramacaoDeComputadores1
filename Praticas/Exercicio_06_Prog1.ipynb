# Exercícios Práticos: O Poder das Funções (Ensinando Novos Truques ao Computador)

Até agora, nós usamos "**comandos prontos**" do Python, como o **print**() e o **input**(). Hoje, vocês darão o maior salto na programação: vão criar os seus **próprios comandos**.
Uma função é como uma "**mini-máquina**" ou uma "**receita**" que você ensina ao computador **uma** única vez e pode usar (chamar) **quantas vezes quiser**, apenas dizendo o **nome** dela.

Para **criar** uma função, usamos a palavra **def** (de definir), damos um **nome** a ela, colocamos **parênteses** () e os dois pontos **:**. Tudo o que estiver com **recuo a direita** (Indentação) faz parte da função. Lembre-se: criar a função **não** faz ela **rodar** na hora. **Depois** de criá-la, você precisa **chamá-la** pelo nome na linha de baixo (sem o recuo) para testar!

## Nível 3: O Comando return (A Função que Devolve a Resposta)
