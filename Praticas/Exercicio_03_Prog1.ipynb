Como funcionará a atividade?
Agora que o computador já sabe te "ouvir" **(input())** e "falar" **(print())**, você vai construir pequenos programas completos. Em cada exercício, você deverá:

1. Criar **variáveis** para receber os dados digitados pelo usuário (lembrando de usar int() ou float() quando for número).

2. Fazer o **cálculo** ou a **comparação lógica** exigida.

3. Mostrar o **resultado** final na tela de forma clara.

###Exemplo Inicial:

Um professor quer saber se a sala tem capacidade para seus alunos. Ele precisa que o sistema pergunte quantos alunos estão matriculados e quantas cadeiras existem na sala, e depois mostre se a quantidade de cadeiras é suficiente (Verdadeiro ou Falso).

O que você deve digitar no Colab:

```python
print("Digite o número de alunos: ")
alunos = int(input())
print("Digite o número de cadeiras: ")
cadeiras = int(input())
suficiente = cadeiras >= alunos
print("Tem cadeiras suficientes para todos?")
print(suficiente)
```
