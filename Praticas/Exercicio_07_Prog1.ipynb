##Questão 1: Impressão Organizada de Etapas

Crie uma função em Python chamada imprimir_etapas(total_etapas) que:

**Parâmetro**: Recebe um número inteiro total_etapas.

**Processamento**: Utilizando o laço while, a função deve imprimir na tela uma mensagem numerada para cada etapa do processo, do passo 1 até o limite informado.

**Retorno**: Nenhum (None).

**Exemplo de Chamada**:imprimir_etapas(3)

**Saída Esperada**:

    [ETAPA 1] Verificação realizada com sucesso.
    [ETAPA 2] Verificação realizada com sucesso.
    [ETAPA 3] Verificação realizada com sucesso.
