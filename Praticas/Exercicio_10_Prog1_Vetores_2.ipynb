#Lista de Exercícios: Vetores e Laços (while)

**Todos** os exercícios devem ser resolvidos obrigatoriamente **utilizando a estrutura de repetição while** para **manipular os vetores**.


##1 Filtro de Resistência Mecânica

Escreva um programa em Python que solicite ao operador a **quantidade de corpos de prova** testados em um ensaio de tração.

Crie um vetor zerado e utilize o laço while para **ler** a tensão de ruptura (em MPa) de **cada amostra**.

Utilizando **outro laço while**, **percorra o vetor** preenchido para calcular a **tensão média do lote** e **contar** quantas amostras apresentaram **resistência abaixo do limite mínimo** de $400.0\text{ MPa}$.

**Imprima** a **média calculada** e o **total de peças reprovadas**.

##2 Localização da Temperatura Máxima no Lingotamento

Um sistema de sensores registra a temperatura em vários pontos de uma linha de lingotamento contínuo.

Peça ao usuário a **quantidade de sensores ativos** e **preencha o vetor de temperaturas** via input() usando o laço while.

Utilizando um segundo laço while, **percorra o vetor** para encontrar qual foi a **maior temperatura registrada** e em **qual posição** (índice) ela se encontra.

**Imprima** o valor da **maior temperatura** e o **índice correspondente**.

##3 Correção de Perda de Massa por Queima

Durante a **fusão de sucata** em um forno a arco elétrico, estima-se uma perda de massa de $5\%$ por oxidação (queima).

**Pergunte ao usuário** o **número de caçambas** de sucata carregadas.

**Preencha um vetor** com o **peso inicial** (em toneladas) de cada caçamba usando o laço while.

Utilize o laço while para **atualizar cada posição do vetor** com o **peso líquido** estimado **após a fusão** ($\text{peso\_liquido} = \text{peso\_inicial} \cdot 0.95$).

**Imprima** o **vetor** com os **pesos devidamente atualizados**.

##4 Comparativo de Composição Química (Planejado vs. Medido)

Uma corrida de liga precisa **comparar** o teor de manganês ($\%Mn$) **planejado** com o **medido** em varias amostras.

Peça ao operador a **quantidade de amostras**.

**Preencha um vetor** medidos com os **valores digitados** pelo usuário via laço while.

Utilizando o laço while, **percorra o vetor e verifique** amostra por amostra: se **o teor medido** estiver dentro da **faixa aceitável** (entre $1.20\%$ e $1.40\%$), exiba "Amostra X: Conforme"; caso contrário, exiba "Amostra X: Fora da faixa".

##5 Relatório Inverso de Resfriamento

Para auditar o **histórico térmico** de uma barra de aço, o engenheiro precisa analisar as taxas de resfriamento da última etapa para a primeira.

**Solicite ao usuário** a **quantidade de etapas registradas**.

**Preencha o vetor** com as **taxas de resfriamento** ($^\circ\text{C/s}$) de cada etapa usando o laço while.

Utilizando outro laço while, **imprima os elementos do vetor** na ordem inversa (do último índice até o índice 0).

##6 Separação de Sucata e Material Aprovado

Em um laminador, o operador registra a massa (em kg) de cada placa produzida e se a peça foi aprovada ou descartada como sucata.

Pergunte a **quantidade de placas** produzidas na jornada.

**Preencha** um vetor com as **massas das placas** e **outro vetor** paralelo **com a situação de cada uma** (1 para aprovada, 0 para sucata) usando o laço while.

Percorra os vetores com o laço while para **calcular** e **exibir** a **massa total acumulada** apenas **das peças descartadas** como sucata.

##7 Desvio em Relação ao Tamanho Médio de Grão

No controle microestrutural, mede-se o tamanho médio de grão (em $\mu\text{m}$) de $N$ regiões de uma amostra.Preencha o vetor com as medições digitadas pelo usuário utilizando o laço while.Em um laço while, calcule a média aritmética de todas as medições.Em um novo laço while, percorra o vetor e imprima o desvio de cada medição em relação à média ($\text{desvio} = \text{valor} - \text{media}$).

##8 Busca Sequencial de Lote Fora de Norma

Escreva um programa para localizar a primeira ocorrência de um teor crítico de fósforo ($\%P$) em um lote de corridas.Solicite ao operador o número de corridas e preencha o vetor com os teores de fósforo via laço while.Solicite ao usuário um teor limite de tolerância para a busca.Utilizando o laço while, percorra o vetor e identifique o índice da primeira corrida que ultrapassou o limite informado, encerrando a busca imediatamente ao encontrar. Caso nenhuma ultrapasse, exiba uma mensagem informando que todas as corridas estão seguras.
