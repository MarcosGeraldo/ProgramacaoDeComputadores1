## Nível 4: Múltiplas Decisões (O if cumulativo)
Nesta etapa, você usará mais de um if independente. O computador vai ler o primeiro, depois vai ler o segundo, e as mudanças podem se acumular.

9. Perfil de Risco (Seguradora)
    Uma seguradora avalia o risco de um cliente. Todo cliente começa com 0 pontos de risco. A seguradora faz duas perguntas independentes:

    O cliente tem menos de 25 anos? (Se sim, ganha 10 pontos de risco).

    O cliente já sofreu acidentes antes? (Se sim, ganha 20 pontos de risco).

    Tarefa:

    1. Crie a variável pontos = 0.

    2. Peça a idade e o número de acidentes passados.

    3. Crie o primeiro if: se idade menor que 25, some 10 aos pontos.

    4. Crie o segundo if (encostado na esquerda, fora do primeiro): se o número de acidentes for maior que 0, some 20 aos pontos.

    5. No final, imprima o total de pontos de risco do cliente.

10. O Carrinho de Compras Inteligente (O Desafio Final)

    Um site de compras online tem uma política flexível. O frete padrão custa R$ 30,00. O sistema analisa os produtos comprados e aplica regras na seguinte ordem:

    Regra 1: Se o valor dos produtos passar de R$ 150,00, o cliente ganha Frete Grátis (o frete zera).

    Regra 2: Se o valor dos produtos passar de R$ 400,00, além do frete grátis, o cliente ganha R$ 50,00 de desconto direto no valor dos produtos.

    Tarefa:

    1. Peça o valor total dos produtos no carrinho.

    2. Crie uma variável frete = 30.0.

    3. Faça o primeiro if (para a regra 1) para zerar o frete se necessário.

    4. Faça o segundo if (para a regra 2) para abater 50 do valor dos produtos se necessário.

    5. No final, crie uma variável total_a_pagar que soma os produtos e o frete. Imprima o valor final exato que o cliente pagará.
