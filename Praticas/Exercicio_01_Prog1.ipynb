Restrição: Não utilize condições ou perguntas (como "se a caixa rasgar..."). Considere um cenário linear onde tudo funciona perfeitamente.

#**Desafio 2**: Transformação de Informação por Regra Fixa
Enunciado Orientador: Um algoritmo frequentemente precisa receber uma informação em um formato e transformá-la em outro, aplicando uma regra **lógica** ou **matemática imutável**. O computador faz isso o tempo todo para calcular horas, valores e dimensões.

##Problema: O Planejamento de um Cronograma.

Premissa: Um gestor informou que a construção de um galpão durará exatamente 4 semanas e 3 dias. Escreva os passos lógicos que um assistente deve seguir mentalmente para converter esse prazo e descobrir o total exato apenas em dias.

#**Desafio 3:** Bifurcação Condicional Simples (Decisão Binária)
Enunciado Orientador: A realidade exige adaptação. Aqui, o seu algoritmo precisa tomar sua primeira decisão autônoma. Dependendo de uma **condição** específica do ambiente, ele deve seguir por um caminho ou por outro, descartando completamente a rota alternativa.

##Problema: A Triagem de Acesso.

Premissa: Um segurança na porta de um evento corporativo fechado verifica os convidados. Descreva os passos para inspecionar um único visitante que acaba de chegar à porta.

Restrição: Use obrigatoriamente a estrutura "Se... então... caso contrário...". O algoritmo deve autorizar a entrada caso a pessoa possua o crachá oficial, ou negar a entrada caso não possua, instruindo a pessoa a se retirar.

# **Desafio 4:** Análise Condicional Composta (Múltiplas Categorias)
Enunciado Orientador: Muitas vezes, um problema não se resume a um simples "Sim" ou "Não". Ele pode ser fragmentado em três ou mais **categorias** **excludentes**, onde apenas uma regra será aplicada dependendo do cenário. O algoritmo precisa saber **classificar** a informação.

##Problema: A Classificação Logística de uma Carga.

Premissa: Um inspetor avalia caixas antes do transporte. Descreva o passo a passo da inspeção de uma caixa que acaba de ser colocada em uma balança.

Restrição: O algoritmo deve possuir três caminhos diferentes de classificação. Se a caixa pesar até 10 kg, deve receber a etiqueta "Leve". Se pesar mais de 10 kg e até 30 kg, deve receber a etiqueta "Padrão". Se pesar estritamente mais de 30 kg, deve receber a etiqueta "Pesada - Requer Empilhadeira".

# **Desafio 5**: Lógica Relacional com Múltiplas Condições Simultâneas
Enunciado Orientador: As decisões mais **complexas** e **seguras** exigem que várias exigências sejam atendidas ao **mesmo tempo** para que uma ação seja liberada. Se apenas uma das regras falhar, o resultado final muda. Essa é a base dos sistemas de segurança.

##Problema: A Aprovação de um Financiamento.

Premissa: Um analista financeiro avalia o pedido de um cliente. Para que o crédito seja concedido, o cliente precisa cumprir duas regras rigorosamente ao mesmo tempo: ter uma renda mensal superior a 3 mil reais E não possuir nenhuma dívida ativa registrada.

Restrição: Estruture os passos de verificação de forma que o financiamento só seja aprovado se ambas as condições forem verdadeiras juntas. Caso o cliente falhe em qualquer uma delas (ou em ambas), o crédito deve ser imediatamente recusado.
