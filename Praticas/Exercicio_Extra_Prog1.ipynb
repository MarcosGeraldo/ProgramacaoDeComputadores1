#Parte 1: Prática sobre Funções

Um centro de usinagem industrial precisa de um programa para calcular o volume de blocos metálicos padronizados antes do envio para o processo de fundição. Você construirá o código em etapas progressivas.


##Questão 1: Função de Cabeçalho Personalizado
Crie uma função **chamada exibir_cabecalho(nome_empresa)** que:

**Entrada** (Parâmetro): Recebe uma string contendo o nome da empresa (nome_empresa).

**Processamento**: Imprime na tela um painel formatado com linhas organizadoras e o nome da empresa centralizado.

**Retorno**: Nenhum (None).

##Questão 2: Função de Cálculo Geométrico Básico

Crie uma função chamada **calcular_area_base(largura, comprimento)** que:

**Entrada** (Parâmetros): Dois números flutuantes representando a largura e o comprimento da base da peça em centímetros.

**Processamento**: Calcula a área da base retangular utilizando a fórmula:$$\text{Área} = \text{largura} \times \text{comprimento}$$

**Retorno**: Retorna o valor calculado da área (float).

##Questão 3: Função Composta (Reutilização de Função)

Crie uma função chamada **calcular_volume_bloco**(largura, comprimento, espessura) que:

**Entrada** (Parâmetros): Três números flutuantes: largura, comprimento e espessura em centímetros.

**Processamento**: Calcula o volume total da peça.Regra Obrigatória: A função deve obrigatoriamente chamar a função **calcular_area_base** (criada na Questão 2) para obter a área da base antes de multiplicar pela espessura:

$$\text{Volume} = \text{Área da Base} \times \text{espessura}$$

**Retorno**: Retorna o volume total em $\text{cm}^3$ (float).

##Questão 4: Integração no Código Principal (main)

Escreva o bloco principal do programa (fora de qualquer função) para executar a seguinte sequência de ações:

1. Solicitar ao usuário o nome da empresa cliente.
2. Chamar a função exibir_cabecalho passando o nome digitado.
3. Solicitar ao usuário a largura, o comprimento e a espessura da peça (em cm).
4. Chamar a função calcular_volume_bloco repassando essas três medidas.
5. Imprimir o resultado final do volume formatado na tela.

##Questão 6: Potência no $N$-ésimo Minuto de Aquecimento

Em tratamentos térmicos de aços especiais, o reaquecimento de peças metálicas pesadas não pode ocorrer de forma abrupta para evitar trincas e tensões internas.

Para garantir um aquecimento uniforme, a potência **$P_k$** (em kW) aplicada por um forno elétrico é aumentada gradualmente a cada minuto **$k$** segundo a equação:

$$P_k = 50 + 3k$$

Escreva um programa em Python que solicite ao usuário o **minuto desejado** $n$ (inteiro). Utilizando um laço while para avançar minuto a minuto **($k = 1, 2, \dots, n$)**, calcule e imprima na tela todas as potências entre **$1$** e  **$P_n$** aplicada no $n$-ésimo minuto.

##Questão 7: Energia Total Acumulada no Forno

Escreva um programa em Python que leia o **tempo total** de aquecimento **$n$** em minutos (inteiro). Utilizando um laço while, calcule e imprima a **soma acumulada da potência** fornecida pelo forno do $1º$ até o $n$-ésimo minuto:$$E_n = \sum_{k=1}^{n} P_k = P_1 + P_2 + \dots + P_n$$
