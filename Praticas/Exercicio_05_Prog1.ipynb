# Exercícios Práticos: Múltiplas Condições com if-else

##Como funcionará a atividade?

Nestes exercícios, os problemas deixaram de ter apenas uma **regrinha simples**. Agora, o computador precisará avaliar duas, três ou mais **condições** ao mesmo tempo usando Operadores Lógicos (and, or, not) e Operadores Relacionais (>, <, >=, <=, ==, !=).
Lembre-se: em todos os exercícios, você deve usar **obrigatoriamente** a estrutura if-else para criar o "Caminho A" (quando as regras são cumpridas) e o "Caminho B" (quando as regras falham).

Dica: Para perguntas de "**Sim ou Não**", peça para o usuário digitar 1 para Sim(**True**) e 0 para Não(**False**), pois facilita a conversão com int().

10. A Porta da Sala do Servidor (Criptografia e Segurança Máxima)
    A sala principal de servidores trocou seu sistema. Agora, não existe mais uma senha fixa (como 9999). Para evitar invasões, a fechadura eletrônica exige uma "senha dinâmica" baseada em propriedades matemáticas. Para a porta abrir, o funcionário precisa ter nível de acesso maior ou igual a 4 E digitar uma senha que cumpra, simultaneamente, todas as regras abaixo:

    1. Ser estritamente maior que 999 e menor que 10000 (ou seja, ter 4 dígitos).

    2. Ser um número múltiplo de 2, de 6 e de 7 ao mesmo tempo.

    Porém, existe uma regra de exceção máxima: se a pessoa na porta for o CEO da empresa, a catraca detecta o crachá-mestre e abre a porta imediatamente, ignorando completamente qual foi a senha digitada e qual é o nível de acesso.

    Tarefa:

    1. Peça para o usuário digitar a senha (número inteiro).

    2. Peça o nível do crachá (número inteiro de 1 a 5).

    3. Pergunte: "Você é o CEO? (Digite 1 para Sim, 0 para Não)".

    4. Construa a lógica do if. Dica de matemática para o computador: para saber se um número é múltiplo de outro, usamos o operador de módulo (%). Se senha % 7 == 0, significa que é múltiplo de 7.

    4. Agrupe toda a validação do funcionário comum usando parênteses.

    5. Por fim, conecte esse grande bloco de regras com a exceção do CEO usando o operador or.

    6. Se a expressão inteira resultar em Verdadeiro, imprima: "Acesso Concedido ao Data Center".

    7. No else (se falhar em qualquer um dos bloqueios e não for o CEO), imprima: "Acesso Negado! Acionando alarme de segurança em 3, 2, 1...".
