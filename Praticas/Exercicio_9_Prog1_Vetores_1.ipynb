#Parte 1: Operações Básicas com Vetores (Sem Uso de Laços)

##1.1 Cadastro e Exibição de Cargas de Liga

Escreva um programa em Python que crie um vetor com **4 posições** contendo os pesos (em kg) de quatro adições de ferro-manganês em uma corrida: 50.0, 75.5, 60.0 e 40.0.

**Sem utilizar** laços de repetição, **imprima na tela** o peso da primeira adição (índice 0) e da última adição (índice 3).

##1.2 Registro e Média Manual de Teor de Carbono

Escreva um programa em Python que crie um **vetor de 3 posições zeradas** para armazenar os teores de carbono ($\%C$) de três amostras de ferro-gusa.

Solicite ao usuário a **digitação de cada teor individualmente** via input(), guardando os valores **diretamente** nos índices 0, 1 e 2 do vetor.

**Sem utilizar funções prontas ou laços**, calcule e imprima a média aritmética dos teores de carbono somando e dividindo as posições do vetor diretamente.



##1.3 Correção de Medição de Espessura

Dado o vetor contendo as espessuras (em mm) de **três chapas** laminadas: espessuras = [12.5, 10.0, 15.0].

Escreva uma linha de comando para **alterar a espessura da segunda chapa** (índice 1) para 11.2.

**Imprima o vetor atualizado** e exiba o resultado da diferença entre a espessura da última chapa (índice 2) e da primeira chapa (índice 0).


##1.4 Balanço de Massa Individual

Escreva um programa que **crie um vetor** com as massas (em kg) de **três amostras** metálicas colhidas de um lingotamento: amostras = [12.4, 11.8, 12.1].

**Sem usar laços de repetição**, calcule a **massa total acumulada** criando uma variável que receba a adição direta do elemento do índice 0 com o do índice 1 e do índice 2 **Imprima** a **massa total** e o **vetor na tela**.

#Bloco 2: Processamento de Vetores com Laço while

**Regra Obrigatória** para os exercícios 2.1 a 2.4: **Utilize** a estrutura de repetição **while** para **percorrer os vetores**.

##2.1 Preenchimento Dinâmico de Medições

Escreva um programa em Python que **pergunte ao operador** quantas medições de **temperatura** de um forno de recozimento serão realizadas na corrida.

Crie um **vetor zerado** com o tamanho $N$ informado pelo operador.

Utilizando o laço while, **solicite a digitação de cada temperatura** e **armazene-a na posição correspondente do vetor** até preencher todas as posições.

**Imprima o vetor final preenchido**.

##2.2 Varredura e Alarme de Temperatura Limitante

Dado um **vetor** com as temperaturas **registradas em 6 pontos** de um forno: [1120.0, 1150.0, 1280.0, 1190.0, 1310.0, 1140.0].

Escreva um programa que utilize o **laço while** para **percorrer o vetor** e **imprimir o valor de cada temperatura**.

**Caso** a temperatura da posição atual seja estritamente **maior que $1200.0^\circ\text{C}$**, imprima **junto ao valor** a **mensagem** de aviso: "[ALERTA: Superaquecimento]".

##2.3 Somatório e Dureza Média de Lote

Dado o **vetor** com os resultados de **ensaios de dureza **Brinell (HBW) de 5 corpos de prova: [210.0, 195.0, 205.0, 220.0, 190.0].

Escreva um programa que utilize o **laço while** para percorrer o vetor, acumulando a **soma de cada elemento em uma variável.**

Ao** final da repetição**, **calcule** e **imprima** a **dureza média** do lote e a **quantidade total** de corpos de prova testados.

##2.4 Contagem de Rejeição de Silício em Lote

Escreva um programa em Python que **pergunte** ao operador **quantas corridas de aço** foram inspecionadas na jornada.

Crie um vetor zerado com o **tamanho informado pelo operador**.

Utilizando a estrutura de repetição while, **solicite ao usuário** a digitação do **teor de silício ($\%Si$) de cada corrida**, **armazenando os valores no vetor**.

Em seguida, utilize novamente o laço while **para percorrer o vetor preenchido**, **contando** e **exibindo na tela** quantas corridas ficaram **fora da especificação** (teor de silício estritamente maior que $0.50\%$).

Ao **final** da varredura, **exiba** o **total geral de corridas reprovadas** no lote.
